```python
import tiktoken
```

```python
enc = tiktoken.get_encoding("o200k_harmony")
```

```python
with open("/nfs/gpu_trainee/getp02/gpt-oss-copy-2/data/input.txt", "r") as f:
    inputs = f.readlines()[1:]
```

```python
inputs = inputs[:4097]
```

```python
lens = [len(enc.encode(x)) for x in inputs]
```
